**EXTRACT** - Extrair a lista de ID do usuário a partir do arquivo CSV.
Para cada ID faça uma requisição do tipo GET para obter os dados do usuário correspondente.

```python
#Importa a API usada na extração de dados.
sdw_2023_api_url = 'https://sdw-2023-prd.up.railway.app'
```

```python
import pandas as pd

defer = pd.read_csv('SDW2023.csv')
userid = defer['UserID'].tolist()
print(userid)
```

```python
import requests
import json

def get_user(id):
  response = requests.get(f'{sdw_2023_api_url}/user/{id}')
  return response.json() if response.status_code == 200 else None

  users = [user for id in userid if (user := get_user(id)) is not None]

  print(json.dumps(users, indent=2))

```
